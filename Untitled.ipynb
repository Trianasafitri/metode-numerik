```python

import math

a = 25
x = 26
h = x - a
asli = math.sqrt(x)

# Menghitung koefisien Taylor
c = [math.sqrt(a), 1/(2*math.sqrt(a))]
for k in range(2, 5):
    c.append(c[k-1] * (0.5-k+1) / (k*a))

# Menghitung pendekatan dan galat
for n in range(5):
    P = sum(c[k] * h**k for k in range(n+1))
    galat = abs(P - asli)
    print(f"P{n} = {P:.10f}, Galat = {galat:.10e}")

    if galat < 1e-6:
        print("Orde minimal =", n)
        break
```
